```python
import pandas as pd
import matplotlib.pyplot as plt

print("Projeto de Análise de Dados")
```

**Bancos** **de** **Dados**

```python
Dados = {
  "data": [
    "2026-09-01", "2026-09-02", "2026-09-03", "2026-09-04",
    "2026-09-05", "2026-09-06", "2026-09-07", "2026-09-08",
    "2026-09-09", "2026-09-10", "2026-09-11", "2026-09-12",
    "2026-09-13", "2026-09-14", "2026-09-15"

  ],
  "produtos": [
      "Notebook", "Mouse", "Teclado", "Monitor",
      "Mouse", "Headset", "Notebook", "Teclado",
      "Monitor", "Mouse", "Headset", "Notebook",
      "Teclado", "Mouse", "Monitor"
  ],
  "categoria": [
      "Informática", "Periféricos", "Periféricos", "Monitor",
      "Periféricos", "Periféricos", "Informática", "Periféricos",
      "Monitores", "Periféricos", "Periféricos", "Informática",
      "Periféricos", "Periféricos", "Monitores"
  ],
  "quantidade": [
      1,3,2,1,4,2,1,3,
      2,5,3,2,4,6,2
  ],
  "preco_unitario": [
      3500,80,150,1200,
      80,250,3500,150,
      1200,80,250,3500,
      150,80,1200
  ],
  "pagamento": [
      "Pix", "Cartão", "Pix", "Cartão",
      "Pix", "Cartão", "Pix", "Dinheiro",
      "Cartão", "Pix", "Cartão", "Pix",
      "Dinheiro", "Pix", "Cartão"
  ]
}

df = pd.DataFrame(Dados)
df.head()
```

**Limpeza** **dos** **Dados**

```python
df["data"] = pd.to_datetime(df["data"])
df["faturamento"] = df["quantidade"] * df["preco_unitario"]
df.info()

df.isnull().sum()
```

**Faturamento** **Total**

```python
faturamento_total = df["faturamento"].sum()
print(f"Faturamento Total: R${faturamento_total:.2f}")
```

## **Produto** **que** **mais** **vendeu**

```python
vendas_produto = df.groupby("produtos")["quantidade"].sum().sort_values(
    ascending = False)
print(vendas_produto)
```

**Gráfico** **de** **vendas**

```python
vendas_produto.plot(kind = "bar")
plt.title("Quantidade vendida por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade")
plt.show()
```

**Produto** **que** **mais** **faturou**

```python
faturamento_produto = df.groupby("produtos")["faturamento"].sum().sort_values(ascending=False)

print(faturamento_produto)
```

**Gráfico** **do** **produto** **que** **mais** **faturou**

```python
faturamento_produto.plot(kind="bar")

plt.title("Faturamento por produto")
plt.xlabel("Produto")
plt.ylabel("Faturamento (R$)")

plt.show()
```

**Faturamento** **por** **categoria**

```python
faturamento_categoria = df.groupby("categoria")["faturamento"].sum().sort_values(ascending=False)

print(faturamento_categoria)
```

**Gráfico** **do** **faturamento** **por** **categoria**

```python
faturamento_categoria.plot(kind="bar")

plt.title("Faturamento por categoria")
plt.xlabel("Categoria")
plt.ylabel("Faturamento (R$)")

plt.show()
```

**Forma** **de** **pagamento** **mais** **utilizada**

```python
pagamentos = df["pagamento"].value_counts()

print(pagamentos)
```

**Gráfico** **de** **vendas** **por** **forma** **de** **pagamentos**

```python
pagamentos.plot(kind="bar")

plt.title("Vendas por forma de pagamento")
plt.xlabel("Forma de pagamento")
plt.ylabel("Quantidade de vendas")

plt.show()
```

# 📌 **Principais** **insights**

* O produto com maior volume de vendas foi identificado através da análise da quantidade vendida.
* O produto com maior faturamento foi identificado através do cálculo de quantidade × preço unitário.
* A categoria com maior participação no faturamento foi identificada pelo agrupamento das vendas.
